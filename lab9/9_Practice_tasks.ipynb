Let us plot a line chart where:

    ii. y axis will represent height

```python
import matplotlib.pyplot as plt
import pandas as pd
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]
df=pd.DataFrame({"height":height,"weight":weight})
#Set xlabel for the plot
plt.xlabel('Weight in kg')
#Set ylabel for the plot
plt.ylabel('Height in cm')
#Set chart title:
plt.title('Average weight with respect to average height')
#plot using marker'-*' and line colour as green
plt.plot(df.weight,df.height,marker='*',markersize=10,color='green',linewidth=2, linestyle='dashdot')
plt.show()
```

# Exercise 4

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

rng = np.random.default_rng(42)
n = 200

sales_data = pd.DataFrame({
    "date": pd.date_range(
        start="2025-01-01",
        periods=n,
        freq="D"
    ),
    "region": rng.choice(
        ["North", "South", "East", "West"],
        size=n
    ),
    "product": rng.choice(
        ["Laptop", "Tablet", "Phone", "Monitor"],
        size=n
    ),
    "advertising": rng.normal(500, 120, size=n).round(2),
    "units_sold": rng.poisson(20, size=n),
    "customer_rating": rng.normal(4.0, 0.5, size=n).clip(1, 5).round(1)
})

product_price = {
    "Laptop": 1200,
    "Tablet": 600,
    "Phone": 800,
    "Monitor": 400
}

sales_data["price"] = sales_data["product"].map(product_price)

sales_data["sales"] = (
    # Quantity * price
    sales_data["units_sold"] * sales_data["price"]
    + sales_data["advertising"] * 3
    + rng.normal(0, 1500, size=n)
).round(2)

sales_data["expenses"] = (
    sales_data["sales"] * rng.uniform(0.55, 0.85, size=n)
).round(2)

sales_data["profit"] = (
    sales_data["sales"] - sales_data["expenses"]
).round(2)

# Add several unusual observations
sales_data.loc[[15, 75, 150], "sales"] *= 2.5
sales_data.loc[[15, 75, 150], "profit"] *= 2.5

sales_data.head()
```

```python
# ---------- Task 1: prepare the data ----------
# Only numeric columns can be correlated
numeric_columns = [
    "advertising", "units_sold", "price", "sales",
    "expenses", "profit", "customer_rating"
]

# corr() computes Pearson correlation for every pair of columns
# (values from -1 to +1)
correlation_matrix = sales_data[numeric_columns].corr()
correlation_matrix.round(2)
```

# Task 2: Create a four-part dashboard 
Create a 2 × 2 figure: \
A. Scatter plot: advertising and sales \
Create a scatter plot in axes[0, 0]. 

B. Histogram: distribution of profit \
Create a histogram in axes[0, 1]. 

C. Box plot: profit by product \
Create a box plot in axes[1, 0]. 

D. Heatmap: correlations \
Create a heatmap in axes[1, 1]. 


```python
# 2x2 grid of axes; figsize makes the figure big enough for 4 plots
fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# A. Scatter plot: advertising vs sales
# hue = color by product (answers "which product sells most"),
# size = point size by units_sold (what the point size represents)

sns.scatterplot(data=sales_data, x="advertising", y="sales",
                hue="product", size="units_sold", alpha=0.7,
                ax=axes[0, 0])
axes[0, 0].set_title("A. Advertising vs Sales")
axes[0, 0].set_xlabel("Advertising spending")
axes[0, 0].set_ylabel("Sales")

# B. Histogram: distribution of profit
# bins = number of intervals; kde=True adds a smooth density curve
sns.histplot(data=sales_data, x="profit", bins=25, kde=True,
             color="steelblue", ax=axes[0, 1])
# Vertical lines for mean and median show the skewness of the data
axes[0, 1].axvline(sales_data["profit"].mean(), color="red",
                   linestyle="--", label="Mean")
axes[0, 1].axvline(sales_data["profit"].median(), color="green",
                   linestyle="-", label="Median")
axes[0, 1].legend()
axes[0, 1].set_title("B. Distribution of Profit")
axes[0, 1].set_xlabel("Profit")

# C. Box plot: profit for each product
# box = middle 50% (Q1..Q3), center line = median,
# whiskers = normal range, single points = outliers
sns.boxplot(data=sales_data, x="product", y="profit",
            hue="product", legend=False, ax=axes[1, 0])
axes[1, 0].set_title("C. Profit by Product")
axes[1, 0].set_xlabel("Product")
axes[1, 0].set_ylabel("Profit")


# D. Heatmap: correlation matrix
# annot=True writes the numbers in the cells; vmin/vmax fix the color
# scale to the full correlation range; cmap="coolwarm" = red positive,
# blue negative
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm",
            vmin=-1, vmax=1, ax=axes[1, 1])
axes[1, 1].set_title("D. Correlation Matrix")
axes[1, 1].tick_params(axis="x", rotation=45)
```

## Answers

**A. Scatter plot**
- *Relationship?* Practically none (correlation ≈ -0.02): the cloud of points has no visible upward or downward trend. Sales depend mostly on product price and units sold, not on advertising.
- *Highest sales?* Laptop (highest price, 1200) – its points are generally at the top.
- *Unusual observations?* Yes – a few points far above the rest (rows 15, 75, 150, where sales were multiplied by 2.5).
- *Point size?* The number of units sold (`units_sold`).

**B. Histogram**
- *Symmetric or skewed?* Right-skewed (long tail to the right) because of the extreme values.
- *Mean vs median?* Mean is higher than the median, a sign of right skew (the outliers pull the mean up).
- *Extreme values?* Yes, the bars far to the right are the 3 inflated observations.
- *Changing bins?* Few bins hide the shape (too smooth); many bins make it noisy/jagged. The overall picture (right skew) stays the same.

**C. Box plot**
- *Highest median profit?* Laptop.
- *Greatest variability?* Laptop (the tallest box / widest spread), since its values are the largest.
- *Outliers?* Products containing the points marked beyond the whiskers – including the inflated rows; they are visible as single dots (mostly Laptop, plus others that contain rows 15, 75, 150).
- *Elements:* box = middle 50% of data (Q1–Q3); center line = median; whiskers = typical range (up to 1.5·IQR); individual points = outliers.

**D. Heatmap**
- *Strongest positive correlation?* `sales` with `expenses` and `profit` (expenses and profit are calculated from sales), and `price` with `sales`.
- *Weak relationships?* `customer_rating` with all variables and `advertising` with most variables (values near 0).
- *Rating vs sales?* No, correlation is close to 0.
- *Why is the diagonal 1?* Every variable is perfectly correlated with itself.
- *Causation?* No. Correlation only shows that variables move together; it does not prove that one causes the other.
