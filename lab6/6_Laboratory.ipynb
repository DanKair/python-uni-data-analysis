# Exercise Set: Data Loading, Storage, and File Formats

### Part 1 – Loading Data
You are given three files:

sales.csv (comma-separated, 1 million rows)

customers.json (nested JSON with customer info)

transactions.xlsx (Excel file with 3 sheets)

Tasks:
- a. Load all three files into pandas DataFrames.
- b. For the JSON file, normalize it so nested fields (like address.city) become separate columns.
- c. Merge the three DataFrames into a single one on customer_id.
- d. Report the total number of transactions per country.

**What this code does:** It loads the sales CSV, flattens nested customer fields such as `address.country` into DataFrame columns, and combines the three Excel sheets into one transactions table. It then joins the tables by `customer_id` and counts distinct transaction IDs for each customer country. A left join keeps rows from the table on the left even when a matching customer or transaction is missing.

```python
import json

# Load sales records from the CSV file.
sales_df = pd.read_csv("sales.csv")

# Flatten nested customer fields into columns such as address.country.
with open("customers.json", "r", encoding="utf-8") as file:
    customer_data = json.load(file)

# Flatten the nested dicts in json via normalize() and seperate them by dot like object properties    
customers_df = pd.json_normalize(customer_data, sep=".")
customers_df = customers_df.rename(columns={"address.country": "customer_country"})

# Read every Excel sheet and stack the transaction rows into one table.
sheets = pd.read_excel("transactions.xlsx", sheet_name=None) # sheet_name=None makes pandas read all sheets at once

# Use pd.concat to stack all dataframes vertically
# sheets.values() extracts just the dataframes, ignoring the sheet name keys

transactions_df = pd.concat(sheets.values(), ignore_index=True) # ignore_index to avoid index collisions

# Join the datasets using the shared customer_id column.
combined_df = (
    sales_df
    .merge(customers_df, on="customer_id", how="left")
    # suffixes() used to avoid columns names conflict (e.g customers.date and transaction.date) to distinguish each other
    .merge(transactions_df, on="customer_id", how="left", suffixes=("", "_transaction"))
)

# Count unique transaction IDs by the customer's country.
country_transactions = transactions_df.merge(
    customers_df[["customer_id", "customer_country"]],
    on="customer_id",
    how="left",
)
transaction_counts = country_transactions.groupby("customer_country")["transaction_id"].nunique()

print("Transactions per country:")
print(transaction_counts.rename("transaction_count"))

#combined_df.to_csv("combined_data.csv", index=False, encoding="utf-8")
```

### Part 2 – Storage Decisions
You now have a combined DataFrame.
Tasks:
- a. Save the dataset in CSV and Pickle formats.
- b. Measure and compare the file sizes and loading times of each format.
- c. Which format would you recommend for:

    Archiving the dataset for future analysis
    
    Sharing with a colleague who does not use Python
    
    Fastest reloading inside Python

```python
import os
import time

# --- Task a: Save the dataset in CSV and Pickle formats ---
print("Saving files...")
csv_filename = "combined_data.csv"
pickle_filename = "combined_data.pkl"

# Save to CSV (index=False prevents saving the row numbers as an extra column)
combined_df.to_csv(csv_filename, index=False)

# Save to Pickle (a binary format native to Python)
combined_df.to_pickle(pickle_filename)


# --- Task b: Measure and compare file sizes and loading times ---

# 1. Measure File Sizes (in megabytes)
csv_size_mb = os.path.getsize(csv_filename) / (1024 * 1024)
pickle_size_mb = os.path.getsize(pickle_filename) / (1024 * 1024)

print(f"\n--- File Size Comparison ---")
print(f"CSV File Size   : {csv_size_mb:.2f} MB")
print(f"Pickle File Size: {pickle_size_mb:.2f} MB")


# 2. Measure Loading Time for CSV
start_time = time.time()
df_loaded_csv = pd.read_csv(csv_filename)
csv_load_time = time.time() - start_time

# 3. Measure Loading Time for Pickle
start_time = time.time()
df_loaded_pickle = pd.read_pickle(pickle_filename)
pickle_load_time = time.time() - start_time

print(f"\n--- Loading Time Comparison ---")
print(f"CSV Load Time   : {csv_load_time:.4f} seconds")
print(f"Pickle Load Time: {pickle_load_time:.4f} seconds")
```

### Part 3 – Database Integration
Create a SQLite database called company.db.
Tasks:
- a. Write the merged DataFrame into a SQL table transactions.
- b. Run an SQL query (using pd.read_sql) to find the top 5 customers with the highest total purchase value.
- c. Export the SQL query result into both Excel and JSON formats.

```python
import sqlite3
import pandas as pd

# 1. Establish a connection to the SQLite database (creates company.db)
conn = sqlite3.connect("company.db")

# 2. Write the combined DataFrame into a SQL table named 'transactions'
# if_exists="replace" drops the table first if it already exists so we can re-run safely
# index=False prevents writing the dataframe row numbers as a separate column
combined_df.to_sql("transactions", conn, if_exists="replace", index=False)

print("Table 'transactions' successfully written to company.db!")

# Part B

# Write a SQL query to calculate total purchase value per customer and get the top 5
query = """
    SELECT customer_id, SUM(amount) AS total_purchase_value
    FROM transactions
    GROUP BY customer_id
    ORDER BY total_purchase_value DESC
    LIMIT 5
"""

# Execute the query and load results into a new DataFrame
top_customers_df = pd.read_sql(query, conn)

print("Top 5 Customers by Total Purchase Value:")
print(top_customers_df)

# Part C

# Export to Excel
top_customers_df.to_excel("top_customers.xlsx", index=False)

# Export to JSON
top_customers_df.to_json("top_customers.json", orient="records", indent=4)

# Always close the database connection when you're done to free up system resources
conn.close()

print("\nFiles successfully exported to top_customers.xlsx and top_customers.json!")
```

### Part 4 – Advanced Challenge
Sometimes we need to handle corrupt/malformed data. Suppose sales.csv has some missing values and extra delimiters.
Tasks:
- a. Load sales.csv with parameters that handle bad lines gracefully (error_bad_lines, na_values, etc.).
- b. Replace missing values in the amount column with the column median.
- c. Save the cleaned file in Feather format.

```python
import pandas as pd

# --- Task a: Load sales.csv handling bad lines and custom missing values ---
# on_bad_lines="skip" drops malformed rows; na_values catches various representations of missing data
sales_cleaned_df = pd.read_csv(
    "sales.csv",
    on_bad_lines="skip",
    na_values=["", "NA", "N/A", "?", "NULL"]
)

# --- Task b: Convert 'amount' to numeric (in case it loaded as text) and fill missing values ---
# errors="coerce" forces any non-numeric values (like text or corrupted symbols) to become NaN
sales_cleaned_df["amount"] = pd.to_numeric(sales_cleaned_df["amount"], errors="coerce")

# 1. Calculate the median of the 'amount' column (now safely numeric)
amount_median = sales_cleaned_df["amount"].median()

# 2. Fill missing (NaN) values in 'amount' with this median
sales_cleaned_df["amount"] = sales_cleaned_df["amount"].fillna(amount_median)

print(f"Missing values in 'amount' filled with median: {amount_median}")


# --- Task c: Save the cleaned file in Feather format ---
# Note: Feather requires a standard range index, so we reset the index just in case
sales_cleaned_df = sales_cleaned_df.reset_index(drop=True)

feather_filename = "sales_cleaned.feather"
sales_cleaned_df.to_feather(feather_filename)

print(f"Cleaned dataset successfully saved to {feather_filename}!")
```
